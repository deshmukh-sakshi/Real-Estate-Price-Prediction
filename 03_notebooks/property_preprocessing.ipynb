```python
import pandas as pd
import numpy as np
import io
from google.colab import files

# ==============================================================================
# STEP 0: UPLOAD YOUR FILE
# ==============================================================================
print("Please upload your 'properties_processed.csv' file:")
uploaded = files.upload()

# Get the file name (assuming it's the one we expect)
file_name = "Properties.csv"

if file_name not in uploaded:
    print(f"\nError: You uploaded a file, but it was not named '{file_name}'.")
    print("Please rename your file and try again.")
else:
    print(f"\nSuccessfully uploaded {file_name}!")

    # ==============================================================================
    # STEP 1: LOAD THE DATA
    # ==============================================================================
    # Read the uploaded file's bytes into a pandas DataFrame
    df = pd.read_csv(io.BytesIO(uploaded[file_name]))

    print("\n--- DATA PREVIEW (BEFORE CLEANING) ---")
    print(f"Shape: {df.shape}")
    print(f"Missing 'year_built': {df['year_built'].isnull().sum()}")
    print(f"Minimum 'area_sqft': {df['area_sqft'].min()}")
    print(f"Unique cities count: {len(df['city'].unique())}")
    print("----------------------------------------\n")

    # ==============================================================================
    # STEP 2: FIX INVALID DATA (Step 5 in project plan)
    # ==============================================================================
    print("Step 2: Fixing negative 'area_sqft' values...")

    # Find how many negative values exist
    neg_area_count = (df['area_sqft'] < 0).sum()

    # Convert all 'area_sqft' values to their absolute value
    df['area_sqft'] = df['area_sqft'].abs()

    print(f"Fixed {neg_area_count} records with negative 'area_sqft'.")
    print(f"New minimum 'area_sqft': {df['area_sqft'].min()}")

    # ==============================================================================
    # STEP 3: STANDARDIZE CATEGORICAL DATA (Step 4 in project plan)
    # ==============================================================================
    print("\nStep 3: Standardizing 'city' names...")

    # Create the mapping dictionary
    city_map = {
        'Surrat': 'Surat', 'Kalkata': 'Kolkata', 'Chennnai': 'Chennai',
        'Dehli': 'Delhi', 'Mumbaai': 'Mumbai', 'Poona': 'Pune',
        'Jaypur': 'Jaipur', 'Ahemdabad': 'Ahmedabad', 'Hyderbad': 'Hyderabad',
        'Mumbay': 'Mumbai', 'Gurugrm': 'Gurugram', 'Bengluru': 'Bengaluru',
        'Punna': 'Pune', 'Hydrabad': 'Hyderabad', 'Calcutta': 'Kolkata',
        'Gurgaon': 'Gurugram', 'Nodia': 'Noida', 'New Delhi': 'Delhi'
    }

    # Apply the mapping
    df['city'] = df['city'].replace(city_map)

    print(f"Standardized city names. New unique count: {len(df['city'].unique())}")
    print(f"Unique cities now: {df['city'].unique()}")


    # ==============================================================================
    # STEP 4: IMPUTE MISSING VALUES (Step 5 in project plan)
    # ==============================================================================
    print("\nStep 4: Imputing missing values...")

    # 1. Impute 'year_built'
    # Strategy: Use the median year for that specific city.
    print("Imputing 'year_built' with city-wise median...")
    df['year_built'] = df.groupby('city')['year_built'].transform(lambda x: x.fillna(x.median()))

    # Fallback: If a whole city group was NaN, fill with global median
    df['year_built'] = df['year_built'].fillna(df['year_built'].median())

    # 2. Impute 'bedrooms' and 'bathrooms'
    # Strategy: Use the median number for that 'property_type'.
    print("Imputing 'bedrooms'/'bathrooms' with property_type-wise median...")
    df['bedrooms'] = df.groupby('property_type')['bedrooms'].transform(lambda x: x.fillna(x.median()))
    df['bathrooms'] = df.groupby('property_type')['bathrooms'].transform(lambda x: x.fillna(x.median()))

    # Fallback: If any still missing, fill with global median
    df['bedrooms'] = df['bedrooms'].fillna(df['bedrooms'].median())
    df['bathrooms'] = df['bathrooms'].fillna(df['bathrooms'].median())

    print("Missing value imputation complete.")

    # ==============================================================================
    # STEP 5: FINAL VALIDATION (Step 8 in project plan)
    # ==============================================================================
    print("\n--- FINAL VALIDATION (AFTER CLEANING) ---")

    # Check for any remaining nulls
    null_counts = df.isnull().sum()
    print("Remaining missing values (should be 0 for key columns):")
    print(null_counts[null_counts > 0])

    if df['year_built'].isnull().sum() == 0 and df['bedrooms'].isnull().sum() == 0 and df['bathrooms'].isnull().sum() == 0:
        print("\nSUCCESS: All critical missing values have been handled.")
    else:
        print("\nWARNING: Some missing values remain. Review fallbacks.")

    if df['area_sqft'].min() > 0:
        print("SUCCESS: No negative 'area_sqft' values remain.")
    else:
        print("WARNING: Negative 'area_sqft' values still exist.")

    print("----------------------------------------------\n")

    # ==============================================================================
    # STEP 6: EXPORT THE CLEANED FILE
    # ==============================================================================
    output_filename = "properties_CLEANED.csv"

    # Convert DataFrame to CSV
    df.to_csv(output_filename, index=False)

    print(f"SUCCESS: Preprocessing complete.")
    print(f"Your cleaned file '{output_filename}' is ready.")

    # Trigger browser download
    files.download(output_filename)
```

```python
import pandas as pd
import numpy as np
import io
from google.colab import files

# ==============================================================================
# STEP 0: UPLOAD YOUR FILE
# ==============================================================================
print("Please upload your 'PropertyDetails.csv' file:")
uploaded = files.upload()

# Get the file name
file_name = "PropertyDetails.csv"

if file_name not in uploaded:
    print(f"\nError: You uploaded a file, but it was not named '{file_name}'.")
    print("Please rename your file and try again.")
else:
    print(f"\nSuccessfully uploaded {file_name}!")

    # ==============================================================================
    # STEP 1: LOAD THE DATA
    # ==============================================================================
    # Read the uploaded file's bytes into a pandas DataFrame
    df = pd.read_csv(io.BytesIO(uploaded[file_name]))

    print("\n--- DATA PREVIEW (BEFORE CLEANING) ---")
    print(f"Shape: {df.shape}")
    print(f"Missing 'hoa_fee': {df['hoa_fee'].isnull().sum()}")
    print(f"Amenities column (example): {df['amenities'].iloc[0]}")
    print("----------------------------------------\n")

    # ==============================================================================
    # STEP 2: IMPUTE MISSING 'hoa_fee' (Step 5 in project plan)
    # ==============================================================================
    print("Step 2: Imputing missing 'hoa_fee' values...")

    # Get the number of missing values before imputation
    missing_hoa_count = df['hoa_fee'].isnull().sum()

    # Calculate the median
    hoa_median = df['hoa_fee'].median()

    # Impute missing values with the median
    df['hoa_fee'] = df['hoa_fee'].fillna(hoa_median)

    print(f"Imputed {missing_hoa_count} missing 'hoa_fee' values with median ({hoa_median}).")

    # ==============================================================================
    # STEP 3: TRANSFORM 'amenities' (Step 7 in project plan)
    # ==============================================================================
    print("\nStep 3: Transforming 'amenities' column into one-hot encoded flags...")

    # Handle any potential rows where the 'amenities' cell itself is missing
    df['amenities'] = df['amenities'].fillna('None')

    # Create the one-hot encoded (dummy) columns
    # This splits the string by ', ' and creates a new column for each unique amenity
    amenity_dummies = df['amenities'].str.get_dummies(sep=', ')

    # Add a prefix to the new columns for clarity and to avoid name conflicts
    amenity_dummies = amenity_dummies.add_prefix('amenity_')

    print(f"Created {len(amenity_dummies.columns)} new amenity flag columns.")

    # Join the new flag columns back to the original DataFrame
    df = pd.concat([df, amenity_dummies], axis=1)

    # Drop the original 'amenities' column as it's no longer needed
    df = df.drop('amenities', axis=1)
    print("Dropped original 'amenities' column.")

    # ==============================================================================
    # STEP 4: FINAL VALIDATION (Step 8 in project plan)
    # ==============================================================================
    print("\n--- FINAL VALIDATION (AFTER CLEANING) ---")
    print(f"New Shape: {df.shape}")
    print(f"Remaining 'hoa_fee' nulls: {df['hoa_fee'].isnull().sum()}")
    print("\nNew 'amenity' columns created (showing first 5):")

    # Get a list of the new columns
    new_amenity_cols = [col for col in df.columns if col.startswith('amenity_')]
    print(new_amenity_cols[:5])
    print("----------------------------------------------\n")

    # ==============================================================================
    # STEP 5: EXPORT THE CLEANED FILE
    # ==============================================================================
    output_filename = "PropertyDetails_CLEANED.csv"

    # Convert DataFrame to CSV
    df.to_csv(output_filename, index=False)

    print(f"SUCCESS: Preprocessing complete.")
    print(f"Your cleaned file '{output_filename}' is ready.")

    # Trigger browser download
    files.download(output_filename)
```
